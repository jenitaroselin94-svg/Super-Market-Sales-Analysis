```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("SuperMarket Analysis.csv")
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Month'] = df['Date'].dt.month_name()
```

```python
print(df.columns)
```

```python
product_sales = df.groupby('Product line')['Sales'].sum()

print(product_sales)
```

```python
import matplotlib.pyplot as plt

colors = ['red', 'blue', 'green', 'orange', 'purple', 'pink', 'brown', 'cyan']

product_sales.sort_values().plot(
    kind='barh',
    figsize=(10,5),
    color=colors
)

plt.title("Highest Sales by Product Line")
plt.xlabel("Sales")
plt.ylabel("Product Line")

plt.show()
```

```python
monthly_revenue = df.groupby('Month')['Sales'].sum()

print(monthly_revenue)
```

```python
monthly_revenue.plot(
    kind='line',
    marker='o',
    figsize=(8,5)
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("SuperMarket Analysis.csv")


print(df.head())


print(df.isnull().sum())


df['Date'] = pd.to_datetime(df['Date'])


df['Month'] = df['Date'].dt.month_name()


print(df.columns)


product_sales = df.groupby('Product line')['Sales'].sum()

print(product_sales)


colors = ['red', 'blue', 'green', 'orange', 'purple', 'pink', 'brown', 'cyan']

product_sales.sort_values().plot(
    kind='barh',
    figsize=(10,5),
    color=colors
)

plt.title("Highest Sales by Product Line")
plt.xlabel("Sales")
plt.ylabel("Product Line")
plt.show()


monthly_revenue = df.groupby('Month')['Sales'].sum()

print(monthly_revenue)


monthly_revenue.plot(
    kind='line',
    marker='o',
    figsize=(8,5),
    color='purple'
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)

plt.show()
```
